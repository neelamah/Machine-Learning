# XGBoost Cheat Sheet

## 1. What is XGBoost?
- Extreme Gradient Boosting (XGBoost) is a **boosting algorithm** for regression & classification.
- Builds **trees sequentially**, each correcting previous errors.
- Features:
  - Gradient boosting framework
  - Regularization (`lambda`, `alpha`) to reduce overfitting
  - Handles missing values
  - Fast & supports parallel computation

---

## 2. Workflow / Steps
1. Prepare & clean data (features X, target y)  
2. Split data (train/test)  
3. Initialize prediction:  
   \[
   \hat{y}_0 = \text{mean of target (regression)}
   \]  
4. Train tree on **residuals** (negative gradient)  
5. Update predictions:  
   \[
   \hat{y} = \hat{y}_{old} + \eta f_k(x)
   \]  
6. Evaluate: RMSE (regression), Accuracy/F1 (classification)  
7. Analyze feature importance

---

## 3. Key Formulas

**Gradient (Residual):**  
\[
g_i = \frac{\partial L}{\partial \hat{y}_i}
\]

**Leaf Value:**  
\[
w_j = - \frac{\sum_{i \in leaf_j} g_i}{\sum_{i \in leaf_j} h_i + \lambda}
\]

**Final Prediction (Regression):**  
\[
\hat{y} = \hat{y}_0 + \eta \sum_{k=1}^{K} f_k(x)
\]

**Final Prediction (Binary Classification):**  
\[
\hat{y} = \frac{1}{1 + e^{-(\hat{y}_0 + \eta \sum f_k(x))}}
\]

```python
# XGBoost Classification Example

import pandas as pd
from xgboost import XGBClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, confusion_matrix

# --- 1. Create Sample Dataset ---
# Predict if a person buys a product based on Age and Income
data = pd.DataFrame({
    'Age': [22, 25, 47, 52, 46, 56, 55, 60],
    'Income': [15, 20, 35, 40, 36, 45, 50, 48],
    'Bought': [0, 0, 1, 1, 1, 1, 1, 1]  # 0 = No, 1 = Yes
})

X = data[['Age', 'Income']]
y = data['Bought']

# --- 2. Split Dataset ---
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)

# --- 3. Train XGBoost Classifier ---
model = XGBClassifier(objective='binary:logistic', n_estimators=5, max_depth=2, learning_rate=0.5)
model.fit(X_train, y_train)

# --- 4. Make Predictions ---
y_pred = model.predict(X_test)
y_prob = model.predict_proba(X_test)[:,1]  # Predicted probabilities for class 1

# --- 5. Evaluate Model ---
accuracy = accuracy_score(y_test, y_pred)
cm = confusion_matrix(y_test, y_pred)

print("Predictions:", y_pred)
print("Probabilities:", y_prob)
print("Accuracy:", accuracy)
print("Confusion Matrix:\n", cm)
```
