Out-of-Bag (OOB)

OOB = “the data not seen by a specific tree during training,” used for error estimation.

    When training a Random Forest, each decision tree is trained on a bootstrap sample of the training data.

    Bootstrap sample: randomly sampled with replacement from the dataset.

    Some samples are not included in this bootstrap sample. These are called Out-of-Bag (OOB) samples for that tree.

    OOB samples act like a built-in test set for evaluating the performance of each tree without needing a separate validation set.

Advantages:

    Provides an unbiased estimate of model performance.

    Reduces the need for cross-validation.

    Efficient for large datasets.

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.datasets import load_iris
from sklearn.metrics import accuracy_score

# Load dataset
iris = load_iris()
X, y = iris.data, iris.target

# Initialize Random Forest with OOB estimation
rf = RandomForestClassifier(n_estimators=100, oob_score=True, random_state=42)

# Train the model
rf.fit(X, y)

# OOB score
print("OOB Score:", rf.oob_score_)

# Prediction (optional)
y_pred = rf.predict(X)
print("Accuracy on full dataset:", accuracy_score(y, y_pred))
```
